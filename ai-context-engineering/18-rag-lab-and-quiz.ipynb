```python
# --- block 3 ---
def hit_rate(eval_set, k=4):
    hits = 0
    for q, gold_chunk_id in eval_set:
        retrieved_ids = [cid for cid, _ in retrieve_ids(embed(q), k=k)]
        hits += gold_chunk_id in retrieved_ids     # was the RIGHT chunk fetched?
    return hits / len(eval_set)
```
