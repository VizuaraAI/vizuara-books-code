```python
# --- block 2 ---
import numpy as np

def causal_mask(L):
    # 1 = keep (allowed), 0 = block (future)
    return np.tril(np.ones((L, L)))      # lower triangle incl. diagonal

def masked_scores(scores):
    L = scores.shape[0]
    mask = causal_mask(L)
    # where mask is 0 (the future), overwrite the score with -inf
    return np.where(mask == 1, scores, -np.inf)
```
