```python
# --- block 6 ---
def maximum(nums):
    best = nums[0]            # assume the first is largest
    for x in nums[1:]:        # loop over the rest
        if x > best:          # decide: is this one bigger?
            best = x          # update the running best
    return best              # hand back the answer

maximum([3, 1, 4, 1, 5])     # → 5
```
