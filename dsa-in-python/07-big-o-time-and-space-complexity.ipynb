```python
# --- block 2 ---
def total(nums):
    s = 0                 # one extra variable, regardless of n
    for x in nums:
        s += x
    return s              # O(n) time, O(1) space
```

```python
# --- block 3 ---
def doubled(nums):
    out = []              # a new list that will hold n elements
    for x in nums:
        out.append(x * 2)
    return out            # O(n) time, O(n) space
```
