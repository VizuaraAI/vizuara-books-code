```python
# --- block 1 ---
def transpose(M):
    n = len(M)
    for i in range(n):
        for j in range(i + 1, n):        # upper triangle only: j > i
            M[i][j], M[j][i] = M[j][i], M[i][j]
    return M
```

```python
# --- block 2 ---
def rotate_clockwise(M):
    n = len(M)
    # 1. transpose: swap across the main diagonal
    for i in range(n):
        for j in range(i + 1, n):
            M[i][j], M[j][i] = M[j][i], M[i][j]
    # 2. reverse each row in place
    for row in M:
        row.reverse()
    return M
```
