```python
# --- block 1 ---
def all_substrings(s):
    result = []
    n = len(s)
    for i in range(n):            # start index
        for j in range(i, n):     # end index, never before the start
            result.append(s[i:j+1])
    return result

all_substrings("abcd")
# ['a', 'ab', 'abc', 'abcd',
#  'b', 'bc', 'bcd',
#  'c', 'cd',
#  'd']
```

```python
# --- block 2 ---
def reverse(s):
    return s[::-1]

reverse("abcd")   # 'dcba'
```

```python
# --- block 3 ---
def reverse_slow(s):
    result = ""
    for ch in s:
        result = ch + result      # prepend each character... looks innocent
    return result
```
