```python
# --- block 1 ---
def count_distinct_slow(s):
    seen = []
    for ch in s:
        if ch not in seen:      # scans the whole list — O(k)
            seen.append(ch)
    return len(seen)
```

```python
# --- block 2 ---
def count_distinct(s):
    seen = set()
    for ch in s:
        seen.add(ch)        # duplicates are ignored automatically
    return len(seen)
```
