```python
# --- block 1 ---
def count_ones(x):
    count = 0
    while x:
        x &= x - 1      # clear the lowest set bit
        count += 1
    return count
```

```python
# --- block 2 ---
def count_bits_brute(n):
    return [count_ones(i) for i in range(0, n + 1)]
```

```python
# --- block 3 ---
def count_bits(n):
    bits = [0] * (n + 1)          # bits[0] = 0 is our base case
    for i in range(1, n + 1):
        bits[i] = bits[i >> 1] + (i & 1)
    return bits
```
