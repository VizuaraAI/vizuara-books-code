```python
# --- block 2 ---
class Stack:
    def __init__(self):
        self._items = []

    def push(self, x):
        self._items.append(x)          # O(1)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._items.pop()       # O(1)

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self._items[-1]         # O(1)

    def is_empty(self):
        return len(self._items) == 0   # O(1)

    def __len__(self):
        return len(self._items)
```
