```python
# --- block 2 ---
def daily_temperatures(T):
    n = len(T)
    answer = [0] * n
    stack = []                      # holds indices; temps decrease bottom→top
    for i in range(n):
        while stack and T[i] > T[stack[-1]]:
            j = stack.pop()         # day j finally has a warmer day
            answer[j] = i - j       # ...and it is exactly (i - j) days out
        stack.append(i)
    return answer
```
