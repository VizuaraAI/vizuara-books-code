```python
# --- block 4 ---
def min_platforms(arrivals, departures):
    arrivals.sort()
    departures.sort()
    n = len(arrivals)

    i = j = 0
    platforms = 0
    max_platforms = 0

    while i < n and j < n:
        if arrivals[i] <= departures[j]:   # next event is an arrival
            platforms += 1
            max_platforms = max(max_platforms, platforms)
            i += 1
        else:                              # next event is a departure
            platforms -= 1
            j += 1

    return max_platforms
```
