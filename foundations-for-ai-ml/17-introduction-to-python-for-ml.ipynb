```python
# --- block 6 ---
def mse(predictions, targets):
    """Mean squared error: average of squared prediction errors."""
    total = 0.0
    for p, t in zip(predictions, targets):   # march in lockstep
        error = p - t
        total += error ** 2                  # ** is exponentiation
    return total / len(predictions)
```
