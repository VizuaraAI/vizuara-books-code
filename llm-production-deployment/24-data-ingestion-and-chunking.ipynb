```python
# --- block 1 ---
def chunk_fixed(text, size=200, overlap=0):
    words = text.split()
    chunks, start = [], 0
    step = size - overlap                 # how far the window advances
    while start < len(words):
        chunk = words[start : start + size]
        chunks.append(" ".join(chunk))
        start += step
    return chunks
```
