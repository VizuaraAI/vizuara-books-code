```python
# --- block 2 ---
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def predict_proba(w, b, x):
    z = w @ x + b            # 1. linear score  (a scalar)
    p = sigmoid(z)           # 2. squash to (0, 1)
    return p                 # 3. P(y=1 | x)

w = np.array([1.5, -2.0])
b = 0.5
x = np.array([2.0, 1.0])

p = predict_proba(w, b, x)   # -> 0.8175...
pred = 1 if p >= 0.5 else 0  # threshold -> class 1
```
