```python
# --- block 1 ---
import numpy as np

def loss(w):
    return (w - 3.0) ** 2

def grad(w):
    return 2.0 * (w - 3.0)     # dJ/dw

def gradient_descent(w, eta=0.1, n_steps=50):
    history = [w]                      # snapshot for plotting
    for _ in range(n_steps):
        w = w - eta * grad(w)          # THE step: downhill
        history.append(w)
    return w, history
```
