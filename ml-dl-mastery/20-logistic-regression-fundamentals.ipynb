```python
# --- block 2 ---
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict_proba(x, w, b):
    z = w @ x + b          # linear score (logit)
    return sigmoid(z)      # squashed into (0, 1) — a probability
```
