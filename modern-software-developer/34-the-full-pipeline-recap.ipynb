```python
# --- block 2 ---
# streak.py
from datetime import date, timedelta

def streak(days, today):
    have = set(days)
    count, cursor = 0, today
    while cursor in have:          # walk backwards while each day is present
        count += 1
        cursor -= timedelta(days=1)
    return count
```
