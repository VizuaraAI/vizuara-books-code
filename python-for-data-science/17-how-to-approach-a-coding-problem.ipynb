```python
# --- block 2 ---
def average(numbers):
    if not numbers:          # empty list is falsy -> guard clause
        return 0
    total = sum(numbers)     # 2. add them all up
    n = len(numbers)         # 3. how many there are
    return total / n         # 4. true division -> a float
```
